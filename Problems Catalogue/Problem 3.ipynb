# Problem 3: Largest Prime Factor

The prime factors of 13195 are 5,7, 13 and 29.

What is the largest prime factor of the number 600851475143?

This problem focuses on another classic mathematical object: the prime numbers.

Finding the prime factors of a number is a classically difficult problem, which underpins the basics of many cryptography algorithms.

Hence, we cannot expect to solve this super efficiently, and we shall have to use the fact that the given number is not enormous.

The solution here is another one that only tests basic programming skills.

The problem here is twofold: we need a number which is both prime, and a factor of the big number (which we will call n for now).

To determine whether or not a number k is prime, one classic way to do so is to go through all the numbers between 1 and sqrt(k); if any of them neatly divide k, then k is not prime, otherwise it is.

This is a relatively easy implication to prove, can you show it mathematically?

Similarly, to figure out whether a number is a factor of n, we can simply loop through all the numbers between 1 and sqrt(n); if one such number, say q, divides n, then we check whether n/q is prime.

Since we loop q from 1 through sqrt(n), n/q is the largest factor we have checked so far, hence finding us the largest prime factor.

We also track which of the q we have checked is the biggest prime, so that if the smallest prime factor is below sqrt(n), we have also found it.

```python
from math import sqrt

def is_prime(n):
    if n < 2:
        return False
    elif n==2:
        return True

    for i in range(2,int(sqrt(n))):
        if n % i == 0:
            return False
    return True

def largest_prime_factor(n):
    max_small_q = 0
    for q in range(1,int(sqrt(n))):
        if n % q == 0:
            print(f"checking prime {int(n/q)}")
            if is_prime(int(n/q)):
                return int(n/q)
            if is_prime(q):
                max_small_q = q
    return max_small_q

print(largest_prime_factor(600851475143))
```
