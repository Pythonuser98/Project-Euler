#Problem 1

If we list all the natural numbers below 10 that are multiples of 3 or 5, we get 3,5,6 and 9. The sum of these multiples is 23.

Find the sum of all the multiples of 3 or 5 below 1000.

This is a basic programming problem. We can solve this issue quite easily by creating a for-loop that loops over all integers from 1 to 1000.
For each new integer, we check whether it is a multiple of 3 or 5; if it is, we add it to our running tally. If not, we do nothing and move on to the next integer.

```python
tally = 0

for n in range(1,1000):
    if n % 3 == 0 or n % 5 == 0:
        tally+=n

print(tally)
```
